```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('ride_hailing_dataset.csv')

# Display the first 5 rows
df.head()
```

```python
# Group by service_type and calculate average fare
avg_fare_service = df.groupby('service_type')['fare'].mean().sort_values(ascending=False)
print(avg_fare_service.round(2))
```

```python
# Count trips per city
city_trips = df['city'].value_counts()
print(city_trips)

print("\nCity with the most trips:", city_trips.idxmax())
```

```python
# Calculate mean trip distance
avg_distance = df['distance_km'].mean()
print(f"Average Distance per Trip: {avg_distance:.2f} km")
```
